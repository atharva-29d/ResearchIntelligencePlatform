```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, precision_score, recall_score , f1_score
```

```python

df = pd.read_csv("../data/processed/processed_papers.csv")
df
```

```python
df.columns
```

```python
x = df['summary']
y = df['category_code']
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size = 0.2 , random_state = 42 , stratify = y)
```

```python
tfidf = TfidfVectorizer(lowercase=True, stop_words = "english", max_features = 30000, ngram_range=(1,2), min_df=1, max_df=1, sublinear_tf=True)
x_train_tfidf = tfidf.fit_transform(x_train)
x_test_tfidf = tfidf.transform(x_test)
```

```python
# Best Parameters: {'svc__class_weight': None, 'tfidf__stop_words': 'english', 'tfidf__sublinear_tf': True}
# Best Parameters: {'svc__class_weight': None, 'tfidf__stop_words': 'english', 'tfidf__sublinear_tf': True}
# Best Parameters: {'tfidf__max_df': 0.9, 'tfidf__max_features': None, 'tfidf__min_df': 1}
# Best Parameters: {'svc__C': 1, 'tfidf__ngram_range': (1, 2)}
#     # "tfidf__ngram_range": [(1,2)],
#     # "tfidf__max_features": [10000, 20000, 30000, None],
#     # "tfidf__min_df": [1, 2, 5],
#     # "tfidf__max_df": [0.90, 0.95, 1.0],
#     # "tfidf__sublinear_tf": [True, False],
#     # "tfidf__stop_words": [None, "english"],
#     TfidfVectorizer(lowercase=True, stop_words = "english", max_features = 30000, ngram_range=(1,2), min_df=2, max_df=0.95, sublinear_tf=True)
#     ("tfidf", TfidfVectorizer(stop_words = "english", ngram_range= (1, 2), max_df = 0.9 , min_df = 1, max_features = None,
#                               sublinear_tf = True , class_weight=None)),
```

```python
model = LogisticRegression(random_state = 42 , class_weight = 'balanced')
model.fit(x_train_tfidf , y_train)

prediction = model.predict(x_test_tfidf)

accuracy = accuracy_score(y_test , prediction)
print(round(accuracy*100, 2),"%")

train_accuracy = model.score(x_train_tfidf , y_train)
test_accuracy = model.score(x_test_tfidf , y_test)
print((round(train_accuracy*100, 2)),(round(test_accuracy*100, 2)))
```
